```python
import pandas as pd
df = pd.read_csv("MOCK_DATA.csv")

df.head()

df.dtypes
```

```python
type_mapping = {
    "int64": "BIGINT",
    "str": "VARCHAR(255)",
    "object": "VARCHAR(255)"
}
```

```python
for col, dtype in df.dtypes.items():
    sql_type = type_mapping.get(str(dtype), "VARCHAR(255)")
    print(f"{col} {sql_type}")
```
